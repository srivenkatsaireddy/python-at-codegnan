```python
# Types of Arguments
# 1. Positional Arguments:
def sub(x:int,y:int)->int:
    c = x - y
    return c
#function calling block
a,b = map(int,input().split())
res = sub(a,b)
print(res)
result = sub(b,a)
print(result)
```

```python
# 2. Keyword Arguments:
def sub(x:int,y:int)->int:
    c = x - y
    return c
#function calling block
a,b = map(int,input().split())
print(sub(x = a, y = b))
print(sub(y = b, x = a))
print(sub(x = b, y = a))
```

```python
# Keyword Arguments:some of two numbers
def add(x:int,y:int)->int:
    c = x + y
    return c
#function calling block
print(sum(x = 8, y = 6))
print(sum(x = 6, y = 8))
```

```python
# Keyword Arguments:some of three numbers
def add(x:int,y:int,z:int)->int:
    c = x + y + z
    return c
#function calling block
print(sum(x = 8, y = 6,z = 4))
print(sum(x = 8, y = 4,z = 6))
```

```python
# Keyword Arguments:some of four numbers
def add(x:int,y:int,z:int,a:int)->int:
    c = x + y + z + a
    return c
#function calling block
print(sum(x = 8, y = 6, z = 3, a = 9))
```

```python
# Keyword Arguments:some of four numbers with default argument examples:
def add(x,y,z = 0,a=0)->int:
    c = x + y + z + a
    return c
#function calling block
print(add(8,6))
print(add(10,20,30))
print(add(1,34))
```

```python
# 3. default Arguments:
def sub(x=9,y=4)->int:
    c = x - y
    return c
#function calling block
a,b = map(int,input().split())
print(sub())
print(sub(y = b, x = a))
```

```python
def add(x,y,z = 1,a=1)->int:
    c = x * y * z * a
    return c
#function calling block
print(add(8,6))
print(add(10,20,30))
print(add(10,39))
```

```python
# 4.Variable Length of Positional Arguments:
def add(a,b = 0, *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s + a + b
print(add(1))
print(add(1,2,3,4,5,6,7))
```

```python
# 4.Variable Length of Positional Arguments:
def add(a,b = 0, *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s + a + b
print(add(1))
print(add(1,2,3,4,5,6,7))
```

```python
# 5.Variable length Keyword arguments:
def details(name,age,*marks,**kwargs):
    print(type(kwargs))
    print(kwargs)
    print(kwargs.keys())
    return name,age,marks,kwargs
print(details("Ravi", 20, 85,95,49 , gender = "male" , loacation = "India"))

    
    
```

```python
# Pass by Value - immutuable data type
def fun(x):
    print(f"Before update Address of x :{id(x)} and value:{x}")
    x = x + 1
    print(f"After update the x value, Address of x :{id(x)} and value: {x}")
a = 5
print(f"Before function Execution address of a:{id(a)} and value: {a}")
print(fun(a))
print(f"After function Execution address of a:{id(a)} and value: {a}")
```

```python
# Pass by Refference - mutuable data type
def fun(x):
    print(f"Before update Address of x :{id(x)} and value:{x}")
    x.append(100)
    print(f"After update the x value, Address of x :{id(x)} and value: {x}")
a = [1,2,3]
print(f"Before function Execution address of a:{id(a)} and value: {a}")
print(fun(a))
print(f"After function Execution address of a:{id(a)} and value: {a}")
```

```python
# Pass by Refference - mutuable data type
def fun(x):
    print(f"Before update Address of x :{id(x)} and value:{x}")
    x = [1,2,3,4,5,6]
    print(f"After update the x value, Address of x :{id(x)} and value: {x}")
a = [1,2,3]
print(f"Before function Execution address of a:{id(a)} and value: {a}")
print(fun(a))
print(f"After function Execution address of a:{id(a)} and value: {a}")
```
