```python
def BinarySearch(arr,i,j,val):
    
    while i <= j:
        mid = (i+j)//2
        if arr[mid] == val:
            return mid
        elif arr[mid] > val:
            return BinarySearch(arr,i,mid-1,val) # j = mid - 1
        elif arr[mid] < val:
            return BinarySearch(arr,mid+1,j,val) # i = mid+1
    return -1
    
# driver code
arr = [1,2,3,5,8,12,20,39,45,100,200]
val = 200
i = 0
j = len(arr)-1
result = BinarySearch(arr,i,j,val)
print(result)
```

```python
def BinarySearch(arr,val):
    i = 0
    j = len(arr)-1
    while i <= j:
        mid = (i+j)//2
        if arr[mid] == val:
            return mid
        elif arr[mid] < val:
            i = mid + 1
        elif arr[mid] > val:
            j = mid - 1
    return -1
arr = [10,20,30,40,50,60,70]
val = 50
print(BinarySearch(arr,val))
```
