```python
# find factor of a number:
n = 10
factors = []
for i in range(1,n+1):
    if n % i == 0:
        factors.append(i)
print(factors)
    
```

```python
# find factor of a number and print number is prime or not prime:
n = 5
factors = []
for i in range(1,n+1):
    if n % i == 0:
        factors.append(i)
print(factors)
if len(factors) == 2:
    print("prime number")
else:
    print("not a prime number")
    
```

```python
# find whether a number is prime or not prime
num = int(input())
for i in range(2, num//2+1):
    if num % i == 0:
        print(f"{num} is not prime")
        break
else:
    print(f"{num} is prime")
```

```python
# find reverse of a string and check it is palindrome or not palindrome
s = input()
rev = " "
for char in s:
    rev = char + rev
print(rev)
if rev == s:
    print(f"{s} is palindrome")
else:
    print(f"{s} is not a palindrome")
    
```
