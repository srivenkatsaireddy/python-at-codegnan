```python
# factorial of a number using recurssion
def fact(a):
    if a == 1 or a == 0:
        return 1
    return a * fact(a - 1)
fact(6)
```

```python
# fibonacii series:
def fun(a):
    if a == 1 or a == 0:
        return 1
    return fun(a-1) + fun(a - 2)
fun(6)
```

```python
# sum of n natural numbers using recursion:
def sum(a):
    if a == 1:
        return 1
    return a + sum(a-1)
sum(4)
```

```python
# product of n natural numbers using recursion:
def sum(a):
    if a == 1:
        return 1
    return a * sum(a-1)
sum(4)
```

```python
# Find binary number for decimal number using recursion:
def binary(a):
    if a == 1:
        return "1"
    if a == 0:
        return "0"
    return binary(a//2) + str(a%2)
binary(2)
    
```

```python
# same question but in normal proceess of decimal number conversion:
num = int(input())
res = ""
while num > 0:
    bit = num % 2 
    num = num // 2
    res = str(bit) + res
print(res)
```

```python
# Square of number using lambda function:
square = lambda x : x ** 2
print(square(5))
```

```python
# sum of two numbers using Lambda function:
square = lambda x,y : x + y
print(square(5,4))
```

```python
# find square of all list of integers using lambda and map():
arr = [1,2,3,4,5]
res = list(map(lambda x : x ** 2, arr))
print(res)
```

```python
def square(x):
    return x ** 2
arr = [1,2,3,4,5]
res = list(map(square, arr))
print(res)
```

```python
# Find even numbers from list of numbers using filter and lambda function:
lst = [1,2,3,4,5,6]
res = list(filter(lambda x : x % 2 == 0, lst))
print(res)
```

```python
def check_even(x):
    if x % 2 ==  0:
        return x
lst = [1,2,3,4,5,6]
res = list(filter(check_even, lst))
print(res)
```
