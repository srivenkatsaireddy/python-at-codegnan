```python
# Exception Handling Block
# Without exception
try:
    a = int(input())
    b = int(input())
    c = a + b
    print(c)
except:
    print("Something wrong in try block")
else:
    print("There is not Exception in try block")
finally:
    print("Program done")
```

```python
# Exception Handling Block
# Exception with an Exception class
try:
    a = 10
    b = 0
    c = a / b
    prit(c)
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception Handling Block
# Define multiple Exceptions with an Exception class
try:
    a = 10
    b = 0
    c = a / b
    prit(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f"We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception Handling Block
# Define multiple Exceptions with an Exception class
try:
    arr = [1,2,3,4]
    c = arr[10]
    prit(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f"We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception Handling Block
# Define multiple Exceptions with an Exception class
try:
    a = 10
    c = len(a)
    prit(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f"We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```
