```python
"""A valid UID must follow the rules below:

It must contain at least  uppercase English alphabet characters.
It must contain at least  digits ( - ).
It should only contain alphanumeric characters ( - ,  -  &  - ).
No character should repeat.
There must be exactly  characters in a valid UID."""
n = input()
upper = 0
digit = 0
visited = set()
valid = True
if len(n) != 10:
    valid = False
for ch in n:
    if not ch.isalnum():
        valid = False
        break
    if ch in visited:
        valid = False
    visited.add(ch)
    if ch.isupper():
        upper += 1
    elif ch.isdigit():
        digit += 1
if valid and upper >2 and digit > 3:
    print("valid")
else:
    print("Invalid")
```
