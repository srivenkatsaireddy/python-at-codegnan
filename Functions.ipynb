```python
#  4 . with return and  with parameters
# sum of two numbers
# function definition block
def addition(x:int,y:int)->int:
    c = x + y
    return c
    print("hi")
# function calling
a, b = 5, 8
res = addition(a,b)
print(res)
```

```python
# 3. with return and without parameters
# sum of two numbers
# function definition block
def addition()->int:
    x , y = 10, 20
    c = x + y
    return c
    print("hi")
# function calling
res = addition()
print(res)
print(addition())
```

```python
# find whether the number is prime number or not using function
def prime_check(n:int)->bool:
    if n < 2:
        return False
    for i in range(2, n//2+1):
        if n % i == 0:
            return False
    return True   
num = int(input())
print(prime_check(num))  
```

```python
# find the prime numbers from 1 to 100:
def prime_check(n:int)->bool:
    if n < 2:
        return False
    for i in range(2, n//2+1):
        if n % i == 0:
            return False
    return True 
    
for j in range(1,100):
    if prime_check(j):
        print(j, end = " ")
num = int(input())
print(prime_check(num))  
     
```

```python
def large(*num):
    largest= num[0]
    for i in num:
        if i > largest:
            largest = i
    print(largest)
large(10,70,30,40)
```
