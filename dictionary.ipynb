# Dictionary

```python
# dictionary ordered collection
details = {"name": "Ravi", "age" : 20, "course":"Python"}
print(details)
```

```python
# dictionary allows list as key but not keys
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details)
```

```python
# if the dictionary contains multiple items with same key ,then it updates first item key value with  latest defined item value
details = {"name": "Ravi", "age" : 20, "course":"Python", "age":34}
print(details)
```

```python
# accessing values by using keys
details = {"name": "Ravi", "age" : 20, "course":"Python", "age":34}
print(details['course'])
print(details["age"])
details['age'] += 1
details['course'] = "java"
print(details)
print(details['age'])
```

```python
details['location']= "India" # it adds new item to dictionary
print(details)
```

```python
# define empty dicitonary
d1 = {}
d2 = dict()
print(type(d1) , type(d2))
```

### Dictionary methods
   - Get value
   - get(key, default_val):It returns the value if key exists in dictionary otherwise 'default_val'
   - update values and Add items to Dictionary
      - update(new dictionary): Dictionary updates with New_dictionary Items
   - Remove items from dictionary
      - pop(key) : It returns value and remove item if key is exists in dictionary otherwise 'keyError'
      - popitem() : It returns tuple of key value pair and removes the last item from dicitionary
   - Get keys, value and Items from dicitionary
       - keys(): It returns list of dictionary keys
       - values(): It returns list of dictionary values
       - items(): It returns list of tuple key-value pairs

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
#print(details['location'])
print(details.get('name'))
print(details.get('location'))
print(details.get('age', 50))
print(details.get('location', "Location key  not exists"))
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
d = {'marks': [85,90.86], 'age':50}
details.update(d)
print(details)
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details.pop('name'))
print(details)
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details.popitem())
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details.keys())
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details.values())
```

```python
details = {"name": "Ravi", "age" : 20, "course":"Python", "marks":[86,49,20], (1,2):10}
print(details.items())
```
