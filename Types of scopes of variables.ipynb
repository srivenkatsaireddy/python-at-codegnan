```python
# Local Scope:
# We can access the Local Scope Variable within the scope,we can't access the global scope
def fun(x):
    x = x + 1 # x is Local Scope
    print(x)
    return x
a = 10 
res = fun(a)
print(x) # we can define it,it is local
```

```python
# Enclosing Scope:
def outer(x):
    y = 10 # Enclose scope
    def inner():
        a = 10 # Local scope
        z = y + a
        print(f"This is local scope:{z}")
    inner()
    y = y + x
    print(f"This is enclosing scope:{y}")
n = 5
outer(n)
```

```python
# global Scope:
x = 10 # Global Scope
def outer():
    y = 20 # Enclosing Scope
    def inner():
        z = 30 # local Scope
        res = z + y + x
        print(f"This is Local Scope: {res}")
    res = y + x  #30
    inner() 
    print(f"This is Enclosing Socpe: {res}")
outer()
```

```python
# Trying to update the global scope variables inside the local scope by using 'global' keyword
x = 10 # Global Scope
def outer():
    global x
    x = x + 1
    print("This is Local Scope val:",x)
outer()
print("This is Global Scope val:",x)
```

```python
# Trying to update the Enclosing scope variables the local scope
x = 10 # Global Scope
def outer():
    y = 20 # Enclosing Scope
    def inner():
        z = 30 # local Scope
        y = y + 1
        print(f"This is Local Scope: {y}")
    inner() 
    print(f"This is Enclosing Socpe: {y}")
outer()
```

```python
# Trying to update the Enclosing scope variables the local scope using 'nonlocal' keyword
x = 10 # Global Scope
def outer():
    y = 20 # Enclosing Scope
    def inner():
        nonlocal y
        z = 30 # local Scope
        y = y + 1
        print(f"This is Local Scope: {y}")
    y = y + x
    inner() 
    y = y + x
    print(f"This is Enclosing Socpe: {y}")
outer()
```
