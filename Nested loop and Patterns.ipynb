```python
# print diagonal number.
n = 4
for i in range(n):
    for j in range(n):
        if j == n - i - 1:
             print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
# print lower triangle from a square:
n = 5
for i in range(n):
    for j in range(n):
        if i >= j:
            print("*", end = " ")
    print()
```

```python
# print hallow square:
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or j == 0 or i == n-1 or j == n-1:
            print("*", end = " ")
        else:
            print(" ",end=" ")
    print(" ")
```
