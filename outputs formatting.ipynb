## Output Formatting

```python
# 1. Old style formatting (using comma)
name = "Ravi"
age = 20
marks = [85, 98, 65, 85, 78, 98]
# My name Ravi and i am a 20 years old
print("My name",name , "and i am a", age ,"years old")
```

```python
# 2. By using Access Specifiers
# integer - %d
# float - %f
# string - %
sname = "Ravi"
age = 20
marks = [85, 98, 65, 85, 78, 98]
# My name Ravi and i am a 20 years old
print("My name %s and i am a %d years old"%(name,age))
```

```python
# 3. Dot Format
sname = "Ravi"
age = 20
marks = [85, 98, 65, 85, 78, 98]
# My name Ravi and i am a 20 years old
print("My name {} and i am a {} years old".format(name,age))
```

```python
# 4. f Format
sname = "Ravi"
age = 20
marks = [85, 98, 65, 85, 78, 98]
percentage = 89.5345
# My name Ravi and i am a 20 years old
print(f"My name {name} and i am a {age} years old and my Marks are {marks} and percentage is {percentage:.3f}")
```
