```python
# reading integer from user
# integer to float, string, bool
n = int(input())
print(type(n), n)
print(float(n))
print(str(n))
print(bool(n))
```

```python
# reading float value from user
# float to int, string, bool
n = float(input())
print(type(n), n)
print(int(n))
print(str(n))
print(bool(n))
```

```python
# reading complex value
real = int(input())
img = int(input())
c = complex(real, img)
print(type(c), c)
```

```python
# reading list of strings
l = list(map(str, input().split()))
print(l)
```

```python
# reading tuple of integers
t = tuple(map(int, input().split()))
print(list(t))
print(set(t))
print(str(t))
```

```python
# reading set of integers
t = set(map(int, input().split()))
print(t)
print(list(t))
print(tuple(t))
print(str(t))
```
