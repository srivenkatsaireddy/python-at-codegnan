```python
n = int(input())
list = []
for i in range(1,n+1):
    list.append(i)
print(list)
```

```python
# List Comprehension:
# syntax: [exp for iterator in iterable if condition]
# return 1-n numbers as list using List Comprehension
n = int(input())
res = [x for x in range(1,n+1)]
print(res)
```

```python
# Find the First n even numbers:
n = int(input())
res = [x for x in range(0,n) if x % 2 == 0]
print(res)
```

```python
# return list even or odd for list of numbers
a = [20,34,33,55,23,54]
res = ["Odd" if x&1 else "Even" for x in a]
print(res)
```

```python
# even - add 1
# odd - add 0.5
a = [20,34,33,55,23,54]
res = [x+0.5 if x&1 else x+1 for x in a]
print(res)
```

```python
# find one dice combinations:
print([x for x in range(1,7)])
```

```python
# find two dice combinations:
print([(i,j) for i in range(1,7) for j in range(1,7)])
```

```python
# find two dices combinations if any dice face is 4
print([(i,j) for i in range(1,7) for j in range(1,7) if i == 4 or j == 4])
```
