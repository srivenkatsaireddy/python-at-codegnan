String Methods
- Case Conversion Methods
   - upper(): it converts  entire string into upper case
   - lower(): it converts entire string into lower case
   - title():It converts every word first character into upper case and remaining into lowercase
   - swapcase():it converts lower into upper and upper to lowercase
   - capitalize():it converts first character of string into uppercase and remaining into lower case
   - split():it returns the list of substring and splits at white spaces

```python
string = "The ocean is blue and sky is BLUE"
lower = string.lower()
print(lower)
print(string)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
print(string.split())
print(string.split("and"))
print(string.split("E"))
print(string.split("T"))
```

## white space remove methods
    - strip(): it removes the white space to both sides of string
    - lstrip(): it removes the white space from left side of string(it removes the leading white spaces)
    - rstrip(): it removes the white space from right side of string(it removes the lagging white spaces)

```python
string = "####codegnan######"
print(string)
print(string.strip("#"))
print(string.lstrip("#"))
print(string.rstrip("#"))
```

```python
string = "   codegnan    "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

# indexing finding methods

```python
# index(sub_string): It returns the first occurrance index of substring string if exists otherwise "valueError"
# find(sub_string): It returns the 
string = "The ocean is blue and sky is blue"
print(string.index("O"))
print(string.index("is"))
print(string.index("is",14)) # It starts checking from 14th index
print(string.index("z")) # substring not found
```

```python
# index(sub_string): It returns the first occurrance index of substring string if exists otherwise "valueError"
# find(sub_string): It returns the 
string = "The ocean is blue and sky is blue"
print(string.find("O"))
print(string.find("is"))
print(string.find("is",14)) # It starts checking from 14th index
print(string.find("z")) # substring not found
```

# replace(old, new):It replace the old_string with new_string in string

```python
string = "The ocean is blue and sky is blue"
print(string)
print(string.replace("blue","red"))
print(string.replace("blue","red",1))
```

# "sep".join(iterable) : It joins the all iterable elements with separator

```python
lst = ['The', 'ocean', 'is', 'blue', 'and', 'sky', 'is', 'BLUE']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

### string checking methods
    - islower(): It returns True when entire string is in lowercase 
    - isupper(): It returns True when entire string is in uppercase
    - istitle(): It returns True when entire string is in title format
    - isalpha(): It returns True when strings contains Alphabets(i.e lower and uppercase)
    - isalnum(): It returns True when strings either Alphabets or digits
    - isdigit(): It returns True when string contains only digits
    - startswith(substring): It returns True when strings startswith the sub_string
    - endswith(substring): It returns True when string endswith the sub_string

```python
print("hello".islower())
print("HELLO".islower())
```

```python
print("HELLO".isupper())
print("hello".isupper())
print("1234".isdigit())
print("12abc".isalnum())
print("sai".isalpha())
print("Sai Reddy".istitle())
print("Hello world".startswith("Hello"))
print("Hello World".endswith("World"))

```

```python
# count(sub_string): It returns count of non overlapping substring occurance
print("ABCCBA".count('D'))
print("ABCBCB".count('BCB'))
```
