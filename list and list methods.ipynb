## List operations
1. list Concatenation (using '+' operator)
2. list Repetation(using '*' operator)

```python
# List Concatenation
l1 = [1,2,3,4]
l2 = [4,5,6]
res = l1 + l2
print(res)
res1 = l1 * 3
print(res1)
print([0] * 10)
print([] * 5)
print(l1 + [10])
print(l1.append(3))
print(l1)
```

## list Methods
  - Element Adding Methods
     - append(item) : It adds item at end of the list
     - extend(iterable): It adds all iterable elements at end of the list
     - inset(pos, item): This Methods adds the item at specified position

```python
lst = [1,2,3,4]
lst.append(10)
print(lst)
lst.append('code')
print(lst)
lst.append([10,20,30,40])
print(lst)
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

# Element remove Methods
    - remove(val): It removes first occurrence of value if values exists in list, otherwise it raises 'ValueError'
    - pop(): It remove and returns the last of element of list
    - pop(pos): It remove and returns the specified position element from list
    - clear(): It removes all element from list
    - del: It python keyword, used to delete values.

```python
lst = [10,20,30,40,20]
lst.remove(20) # it removes the first occurence of 20
#lst.remove(100)
print(lst)
val = lst.pop() # it removes the last element from the list
print(val)
print(lst)
val2 = lst.pop(1) # it removes the first element from the list
print(val2)
print(lst)
del lst[0]
print(lst)
lst.clear()# it deletes all elements from the list
print(lst)
```

## index(val): It returns the index of first occurence of value if value exists in list otherwise valueError
## count(val): It returns the total frequence of value

```python
lst = [1,2,3,4,5]
a = lst.index(4)
print(a)
print(lst.count(2))
print(lst.count(100))
print(lst.index(100))
```

```python
## sort(reverse = false) : It sorts the all list elements in ascending order, if reverse value is True it sorts in descending order.
lst = [5,43,63,6,73,-2,4,-6]
lst.sort()
print(lst)
```

```python
lst = ["sai","hi","hello","dolly"]
lst.sort()
print(lst)
```
