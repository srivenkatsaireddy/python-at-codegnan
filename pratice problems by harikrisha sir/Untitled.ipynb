```python
def binarysearch(nums,target):
    low = 0
    high = len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            return f"Element found at index {mid}"
        elif nums[mid] > target:
            high = mid - 1
        else:
            low = mid + 1
    return "Element not found"
nums = [10,20,30,40,50]
target = 40
result = binarysearch(nums,target)
print(result)
            
```

```python
def Binarysearch(nums,target):
    low = 0
    high = len(nums)-1
    answer = -1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            answer = mid
            high = mid - 1
            return f"First occurrence of {target} is at index {answer}"
        elif nums[mid] > target:
            high = mid - 1
        else:
            low = mid + 1
    return "Element not found"
nums = [10,20,20,20,30,40,50]
target = 20
result = Binarysearch(nums,target)
print(result)
```
