```python
def add(x):
    x=x+1
    yield x
    x=x+2
    yield x
res=add(5)
print(res)
print(next(res))
print(next(res))
print(next(res))
```

```python
def add(x):
    x=x+1
    yield x
    x=x+2
    yield x
res=add(5)
for i in res:
    print(i)
```

```python
def numbers(x):
    res=[]
    for i in range(1,x+1):
        res.append(i)
    return res    

x=int(input())
numbers(x)      
```

```python
def gen(x):
    for i in range(1, x+1):
        yield i 
res = gen(10)
for i in res:
    print(i)
```
