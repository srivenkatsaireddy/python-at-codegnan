```python
 # 1) check whether number is +ve or -ve
a = int(input())
if a >= 0:
    print(f" {a} is positive numbers.")
else:
    print(f" {a} is negative numbers.")
```

```python
a = int(input())
if a <= -1:
    print(f" {a} is negative numbers.")
else:
    print(f" {a} is positive numbers.")
```

```python
# 2) check whether the number is divisible with 3 or not
a = int(input())
if a % 3 == 0:
    print(f"{a} is divisible with 3")
else:
    print(f"{a} is not divisible with 3")
```

```python
# 3) check whether the number is divisible with 3 and 5
a = int(input())
if a % 3 == 0 and a % 5 == 0:
    print(f"{a} is divisible with 3 and 5")
else:
    print(f"{a} is not divisible with 3 and 5")
```

```python
# 4)check whether the number is multiple of 3 and 5
a = int(input())
if a % 3 == 0 and a % 5 == 0:
    print(f"{a} is multiple with 3 and 5")
else:
    print(f"{a} is multiple with 3 and 5")
```

```python
# 5) check whether the number is perfect square or not
a = int(input())
val =  a**0.5 
if val == int(val):
    print(f"The given {a} is perfect square")
else:
    print(f"The given {a} is not perfect square")
```

```python
# 6) Return index of target if target exist in list of integer otherwise -1
a = list(map(int,input().split()))
b = int(input())
if b in a:
    print(a.index(b))
else:
    print("-1")
```

```python
# 7) check whether the number is even or odd without using "else" block
num = int(input())
if num % 2 == 0:
    print(f" {num} is even")
if num % 2 != 0:
    print(f"{num} is Odd")
```

```python
# 8) check whether the number is divisible with 3 = return 'Fizz' 
 #  if 5 = return 'Buzz' 
num = int(input())
if num % 3 == 0:
    print("Fizz")
if num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

```python
# 9) check whether the number is positive , negative , zero
num = int(input())
if num > 0:
    print(f"{num} is positive number")
elif num < 0:
    print(f"{num} is negative number")
else:
    print(f"{num} is zero number")
```

```python
# 9) check whether the number is divisible with 3 = return 'Fizz'
# 5 = return 'Buzz'
# 3 & 5 = "Fizz Buzz'
 # else  = number
num = int(input())
if num % 3 & num % 5 == 0:
    print("Fizz Buzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
```

```python
# 10) 
num = int(input())
if num > 0 and num % 2 == 0:
    print("positive even number")
elif num > 0 and num % 2 == 1:
    print("positive odd number")
elif num < 0 and num % 2 == 0:
    print("negative even number")
elif num < 0 and num % 2 == 1:
    print("negative odd number")
else:
    print("zero")
        
```
