```python
# printing number using nested loops
for i in range(1,3):
    for j in range(1,4):
        print(i,j)
```

```python
# printing '*' in single line
for i in range(1,5):
    for j in range(1,5):
        print("*",end = " ")
```

```python
# square shape '*' pattern:
n = 5
for i in range(n):
     for j in range(n):
         print("*", end = " ")
     print()
```

```python
# square shape '1' pattern:
n = 5
for i in range(n):
     for j in range(n):
         print("1", end = " ")
     print()
```

```python
# square shape with 'numbers' pattern: with formula n * i + j + 1
n = 4
for i in range(n):
    for j in range(n):
         print(n * i + j + 1, end = "\t")
    print()
```

```python
# square shape with 'same number in one row' patter:
n = 4
for i in range(n):
    for j in range(n):
        print(i + 1, end = " ")
    print()
```

```python
# square shape with 'same number in one coloumn' patter:
n = 4
for i in range(n):
    for j in range(n):
        print(j + 1, end = " ")
    print()
```

```python
# print number accordingly.
n = 4
for i in range(n):
    for j in range(n):
        print((i+j) % 2, end = " ")
    print()
```

```python
# print diagonal number.
n = 4
for i in range(n):
    for j in range(n):
        if i == j:
             print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
# print diagonal number.
n = 4
for i in range(n):
    for j in range(n):
        if j == n - i - 1:
             print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
# print diagonal number.
n = 5
for i in range(n):
    for j in range(n):
        if j == n - i - 1 or i == j:
             print("*",end = " ")
        else:
            print(" ",end = " ")
    print()
```

```python
# print lower triangle from a square:
n = 5
for i in range(n):
    for j in range(n):
        if i >= j:
            print("*", end = " ")
    print()
```

```python
# print upper triangle from a square:
n = 5
for i in range(n):
    for j in range(n):
        if i <= j:
            print("*", end = " ")
        else:
            print(" ", end = " ")
    print()
```

```python
# print another shape of traingle from a square
n = 5
for i in range(n):
    for j in range(n):
        if j <= n-i-1:
            print("*", end = " ")
    print()
```

```python
# print another shape of traingle from a square
n = 5
for i in range(n):
    for j in range(n):
        if j >= n-i -1:
            print("*", end = " ")
        else:
            print(" ", end=" ")
    print()
```

```python
# print lower triangle from a square with numbers ascending:
n = 5
k = 1
for i in range(n):
    for j in range(n):
        if i >= j:
            print(k, end = " ")
            k += 1
    print()
```

```python
# print lower triangle from a square with numbers decreasing:
n = 5
k = 15
for i in range(n):
    for j in range(n):
        if i >= j:
            print(k, end = "\t")
            k -= 1
    print()
```

```python
# print hallow square:
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or j == 0 or i == n-1 or j == n-1:
            print("*", end = " ")
        else:
            print(" ",end=" ")
    print(" ")
```

```python
# print hallow right angle triangle:
n = 5
for i in range(n):
    for j in range(n):
        if j == 0 or i == n-1 or i == j:
            print("*",end = " ")
        else:
            print(" ", end = " ")
    print(" ")
            
```

```python
# print 'z' through hallow square:
n = 5
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == n-i-1:
            print("*",end = " ")
        else:
            print(" ", end = " ")
    print(" ")
```

```python
# print 'y' through hallow square:
n = 5
for i in range(n):
    for j in range(n):
        if j == n-i-1  or (i == j and i <= n//2):
            print("*",end = " ")
        else:
            print(" ", end = " ")
    print(" ")
```
