```python
# Method Overriding:
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def details(self):
        return "person class executed"
class Student(Person):
    def __init__(self,name,age,roll_no):
        super().__init__(name,age)
        self.roll_no = roll_no
    def details(self):
        return "student class executed"
c = Student("sai",45,20)
print(c.details())
d = Person("kalyan",56)
print(d.details())
```

```python
# Method Overriding:
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def details(self):
        return "person class executed"
class Student(Person):
    def __init__(self,name,age,roll_no):
        super().__init__(name,age)
        self.roll_no = roll_no
    def details(self):
        print(super().details())
        return "student class executed"
c = Student("sai",45,20)
print(c.details())
#d = Person("kalyan",56)
#print(d.details())
```

```python
# Method Overloading: by default values
class Math:
    def add(self,a,b,c=3,d=7):
        return a + b + c + d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2))
print(m.add(1,2,3))
```

# python directly doesn't support method overloading but we can achieve through the default arguments,arbitary arguments 

```python
# Operator Overloading:
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
p1 = Point(5,3)
print(p1)
```

```python
# Operator Overloading:
class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y
    def __str__(self): #string magic method
        return f" This point class and value is:({self.x},{self.y})"
p1 = Point(5,3)
print(p1)
```

```python
# Operator Overloading:
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # string magic method
        return f"This point class and value is:{self.x}"
    def __add__(self,other):
        return self.x + other.x
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
print(p1 + p2)
```

```python
# Operator Overloading:
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self): # string magic method
        return f"This point class and value is:{self.x}"
    def __add__(self,other):
        return self.x + other.x
    def __sub__(self,other):
        return Point(self.x - other.x)
p1 = Point(5)
p2 = Point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
```
