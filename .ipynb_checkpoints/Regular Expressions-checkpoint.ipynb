```python
import re
```

```python
# match(pattern, string)
string = "The sky is blue and ocean is blue"
res = re.match("[A-Z]",string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern value
```

```python
# match(pattern, string)
string = "The sky is blue and ocean is blue"
res = re.match("[A-Z,a-z]+",string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern value
```

```python
# match(pattern, string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("z",string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern value
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# search(pattern, string)
# search 5 length word
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("[A-Za-z]{5}",string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern value
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# get the words from string which length is morethan 3
# findall(pattern,string)
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[A-Za-z]{3,}",string)
    print(res)
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# validate mobile number 
# all are digits,length of number is 10,starts with 6,7,8,9, Remaining are 0-9
try:
    mobile = input()
    res = re.findall("^[6-9][0-9]{9}",mobile)
    
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# validate pan card number
# it allows upper and digits,length is 10, first 5 are upper ,next 4 are digits, and last one is upper
import re
try:
    pan = input()
    res = re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]$",pan)
    
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
#validate email id
import re
try:
    email = input()
    res = re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9]+\.[a-z]{2,3}",email)
    
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# validate date/month/year:
import re
try:
    date = input()
    res = re.fullmatch(r'^(0?[1-9]|[12][0-9]|3[01])[-/](0?[1-9]|1[0-2])[-/]\d{4}$',date)
    print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is Invalid")
except Exception as e:
    print("There is no matched pattern:",e)
```

```python
# stop watch:
import time
from IPython.display import clear_output
sec_counter = 0
min_counter = 0
hr_counter = 0
while True:
    print(f"{hr_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter += 1
    if sec_counter == 60:
        min_counter += 1
        sec_counter = 0
        if min_counter == 60:
            hour_counter += 1
            min_counter = 0
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait = True)
```

```python
# timer watch:
import time
from IPython.display import clear_output
minutes = int(input("Enter stop watch minutes:"))
sec_counter = 0
min_counter = 0
hr_counter = 0
while min_counter != minutes:
    print(f"{hr_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter += 1
    if sec_counter == 60:
        min_counter += 1
        sec_counter = 0
        if min_counter == 60:
            hour_counter += 1
            min_counter = 0
            if hour_counter == 24:
                hour_counter = 0
    clear_output(wait = True)
```
