```python
# Linear Searching:----Time complexity--o(n)
def Lsearch(arr,val):
    for i in range(len(arr)):
        if arr[i] == val:
            return i
    return -1
#driver code
arr = [10,2,45,3,7,24,90]
val = 7
result = Lsearch(arr,val)
print("value:",val)
print(result)
```

```python
# Binary Searching:-----Time complexity -- O(log n base 2)
def BinarySearch(arr,i,j,val):
    
    while i <= j:
        mid = (i+j)//2
        if arr[mid] == val:
            return mid
        elif arr[mid] > val:
            j = mid-1
        elif arr[mid] < val:
            i = mid+1
    return -1
    
# driver code
arr = [1,2,3,5,8,12,20,39,45,100,200]
val = 200
i = 0
j = len(arr)-1
result = BinarySearch(arr,i,j,val)
print(result)
```

```python
# Binary Searching:-----Time complexity -- O(log n base 2)
def BinarySearch(arr,i,j,val):
    
    while i <= j:
        mid = (i+j)//2
        if arr[mid] == val:
            return mid
        elif arr[mid] > val:
            return BinarySearch(arr,i,mid-1,val) # j = mid - 1
        elif arr[mid] < val:
            return BinarySearch(arr,mid+1,j,val) # i = mid+1
    return -1
    
# driver code
arr = [1,2,3,5,8,12,20,39,45,100,200]
val = 200
i = 0
j = len(arr)-1
result = BinarySearch(arr,i,j,val)
print(result)
```

```python
# Find the infinity index in an array
def BinarySearchElement(arr1,i,j,x):
    if i<=j:
        mid = i+(j-i)//2
        if(arr1[mid] == x):
            return mid
        elif arr1[mid]<x:
            return BinarySearchElement(arr1,mid+1,j,x) # or use i=mid+1 , keep it in mind, use while loop there
        else:
            return BinarySearchElement(arr1,i,mid-1,x) # or use j=mid-1 ,  keep it in mind, use while loop there
    return -1

## Driver Code
arr = [1,32,5,3,-6,34,-50,10,1,-8,12,'inf','inf','inf','inf']
arr1 = []
for i in arr:
    arr1.append(float(i))
x = float('inf')
i = 0
j = len(arr)-1
result = BinarySearchElement(arr1,i,j,x)
print(result)
```
