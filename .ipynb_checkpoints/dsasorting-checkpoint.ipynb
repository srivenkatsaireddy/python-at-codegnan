```python
# BUbble sort:
def Bubblesort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr

arr=[5,6,10,45,11,4,26,1,2,72,8]
Bubblesort(arr)
print(arr)
```

```python
# Insetion sort:
def InsertionSort(arr):
    for i in range(1,len(arr)):
        key = arr[i]
        j = i - 1
        while j >= 0 and key < arr[j]:
            arr[j+1] = arr[j]
            j -= 1
        arr[j+1] = key
    return arr
arr = [7,22,64,12,2,6,9,25,67,1]
result = InsertionSort(arr)
print(result)
```
