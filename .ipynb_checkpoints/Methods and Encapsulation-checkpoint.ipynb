```python
class Accounts:
    name = "venkat"
    def __init__(self,username,balance,password):
        self.username = username
        self.balance = balance
        self.password = password
    
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
            return f"{amount} withdrawal successfully and current {self.balance}"
        return "Insufficient Balance"

    def deposit(self,amount):
        self.balance += amount
        return f"{amount} Deposite successfully and current Balance is {self.balance}"
user = Accounts("venkatsai",45000,"venkat@123")
print(user.deposit(1000))
#user.balance = 10000
print(user.withdraw(5000))

```
