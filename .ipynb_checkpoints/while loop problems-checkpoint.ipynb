```python
# Remove target element from list of integers using for looop
arr = list(map(int,input().split()))
target = int(input())
res = []
for i in arr:
   if i != target:
       res.append(i)
       
print(res)
    
```

```python
# Remove target element from list of integers using while loop
arr = list(map(int, input().split()))
target = int(input())
res = []
n = len(arr)
i = 0
while i < n:
    if arr[i] != target:
        res.append(arr[i])
    i = i + 1
print(res)
```

```python
arr = list(map(int, input().split()))
target = int(input())
n = len(arr)
i = 0
while i < n:
    if arr[i] == target:
        arr.pop(i)
        n -= 1
    else:
        i = i + 1
print(i)
```
