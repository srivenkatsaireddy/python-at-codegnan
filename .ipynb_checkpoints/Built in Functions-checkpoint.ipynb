# Built in functionns: The Functions definition is provided by python, those functions called as Built in Functions

```python
# abs() : It returns the aboslute value
print(abs(20-10))
print(abs(10-20))
```

```python
# round(val, pos) : This function rounds the float value to particular positions
print(round(2.2589,2))
print(round(2.2529,2))
print(round(2.2589))
print(round(2.6589))
```

```python
# ord(char) : It returns the ASCII value of character
# chr(ascii_value) : It returns the character for specified ASCII value
print(ord("A"))
print(ord("a"))
print(chr(65))
```

```python
# sorted(iterable, reverse=False, key = function): It returns new sorted iterable
# sort the fruits based on the length of fruits name
fruits = ['Apple','bannana','kivi','Mango','Water Melon']
res = sorted(fruits)
print(res)
print(sorted(fruits,reverse = True))
print(sorted(fruits,reverse = False))
print(sorted(fruits,reverse = True, key = len))
```

```python
# map(function, iterable)
# Find the square of all list of integers using map function
def square(x):
    return x ** 2
arr = [1,2,3,4,5]
res = list(map(square, arr))
print(res)

        
```

```python
# filter(function, iterable): It filters the iterables, if the function reuturs Non-None those values will added to result
# Find even numbers from list of numbers
def check_even(x):
    if x % 2 ==  0:
        return x
lst = [1,2,3,4,5,6]
res = list(filter(check_even, lst))
print(res)
        
```

```python
# Same filter() question do it with map() function
# Find even numbers from list of numbers
def check_even(x):
    if x % 2 ==  0:
        return x
lst = [1,2,3,4,5,6]
res = list(map(check_even, lst))
print(res)
        
```

```python
# Find the 3 's Multipliers from list of integers
arr = [1,24,54,22,44,51,33,15]
lst = []
for i in arr:
    if i % 3 == 0:
        lst.append(i)
print(lst)
```

```python
def Three_multiplier(x):
    if x % 3 ==  0:
        return x
lst = [1,24,54,22,44,51,33,15]
res = list(filter(Three_multiplier, lst))
print(res)
```

```python
# enumerate(iterable): This function reutrns  the list of tuple of index and item for iterable items
fruits = ['Mango','Apple','bannana','kivi','Water Melon']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind, val in enumerate(fruits):
    print(f"{ind}-{val} and length is {len(val)}")
```

```python
# Tuple unpacking
t = (1,2)
a, b= t
print(a,b)
```

```python
# zip(iterable1,iterable2,......) : It returns all iterables as single iterable of tuples
rool_no = [101,102,103,104,105,106]
names = ['Ravi','Raju','Siva','Rani','Gopi','sai']
percentage = [90,67,96.8,67.65,99.99999,89]
print(list(zip(rool_no,names,percentage)))
```

```python
# number conversion functions
# binary conversion
num = int(input())
res = ""
while num > 0:
    bit = num % 2 
    num = num // 2
    res = str(bit) + res
print(res)
```

```python
# octal conversion
num = int(input())
res = ""
while num > 0:
    bit = num % 8
    num = num // 8
    res = str(bit) + res
print(res)
```

```python
# number conversions using functions:
print(bin(10))
print(bin(10)[2:]) #slicing
print(oct(10))
print(oct(10)[2:]) # slicing
print(hex(11))
print(hex(11).upper()[2:]) # upper and slicing
```

```python
# any() : This Function returns True when any one iterable items return True
# all() : Return True if bool(x) is True for all values x in the iterable.
arr = [1,2,3,4,0]
arr1 = [1,1,2]
print(all(arr1))
print(any(arr))
```

```python
percentage = [34,35,54,65,67]
if all(percentage < "30"):
    print("ALL STUDENTS ARE PASSED")
else:
    print("ALL STUDENTS ARE FAILED")
```
