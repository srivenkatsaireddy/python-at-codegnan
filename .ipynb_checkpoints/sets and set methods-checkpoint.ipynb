## sets

```python
# set is Unordered Collection
s = {10,2,4,9,3, "code", "hello", 2.25,9} # it arranges according to hash values.
print(s)
A = {8,2,9,7,7,77} # returns only unique elements
print(A)
print(len(A))
s = {4,3,True,1} # True and 1 are same, only one item is printed as True
print(s)
```

```python
# set doesn't allow unhassable list 
s = {1,2,3,4,[43,54]}
print(s)
```

```python
# set doesn't allow tuple
s = {1,2,3,4,(43,54)}
print(s)
# we can also add tuple
s = {1,2,3,4}
s.add((5,6,7))
print(s)
```

```python
# define empty set 
s = {}
print(s)
print(type(s))
a = set()
print(a)
print(type(a))
a = set("1,3,3")
print(a)
b = set([1,2,3])
print(b)
c = set((1,2,3))
print(c)
```

# Set Methods
  - Adding Elements To Set
     - add(ele): It adds element to set if element not exists in set, otherwise nothing
     - update(iterable) : It adds all iterable elements to set 
  - Remove elements from Set
     - remove(ele): It removes the element from set if element exists in set otherwise 'keyError'
     - discard(ele): It removes the element from set if element exists in set otherwise nothing
     - pop() : It removes and returns the element from the set


```python
s = {1,3,4,5,6,"set",6}
print(s)
print(s.add(5))
s.add(9)
print(s)
print(len(s))
s.remove(5)
print(s)
s.discard(6)
print(s)
s.discard(9)
print(s)
print(s.pop()) # it also retuns also
print(s)
s.clear()
print(s)
# remove and discard are same but if we perform there is no element remove gives 'keyError' and discard gives nothing.
```

# set operations
  - union(another_sets): It returns the all common elements from all sets
  - intersection(another_sets): It returns the common elements from the all sets
  - difference(another_sets): It returns these set elements those are not lies in another_sets
  - symmetric_difference(another_sets): It returns the Non-common elements from all sets


```python
s1 = {2,3,4,6}
s2 = {2,9,0,3}
print(s1.union(s2))
print(s1.intersection(s2))
print(s1.difference(s2))
print(s2.difference(s1))
print(s1.symmetric_difference(s2))
```

```python
s1 = set()
s2 = {1, 2, 3}
print(s1.intersection(s2))
print(s1.union("4","6"))

print(s1.difference(s2))
print(s2.difference(s1))
print(s1.symmetric_difference(s2))
```

```python
s1 = {2,3,4,6}
s2 = {2,9,0,3}
print(s1 | s2) # union
print(s1 & s2) # intersection
print(s1)
print(s1 - s2) # difference
print(s2 - s1) # difference
print(s1 ^ s2) # symmetric difference
```

```python
s1 = {1,2,3}
s2 = {2,3,4,5}
s1.intersection_update(s2)
print(s1)
print(s2)
s1.difference_update(s2)
print(s1)
print(s2)
s1.symmetric_difference(s2)
print(s1)
print(s2)
```

## Set Checking Methods
  - issubset(another_set)
  - issuperset(another_set)
  - isdisjoint(another_set)

```python
print({1,2,3}.issubset({1,2,3,4,4}))
print({1,2,3,4}.issuperset({1,2}))
print(set().issubset({1}))
print({1,2}.isdisjoint({4,5}))
```
