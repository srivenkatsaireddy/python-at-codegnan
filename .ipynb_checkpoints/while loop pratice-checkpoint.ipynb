```python
# print number from 1 to 10 using while loop
i = 1
while i <= 10:
      print(i)
      i = i + 1 
```

```python
# print number from 10 to 1 in reverse using while loop
i = 10
while i >= 1:
     print(i)
     i = i - 1
```

```python
lst = [2,5,7,34,23]
mini = float("inf")
for num in lst:
    if num < mini:
        mini = num
print(mini)
```

```python
lst = [3,5,35,34,94]
max = float("inf")
for num in lst:
     if num > max:
           max = num
print(max)
```

```python
for r in range(1,4):
    for c in range(1,4):
         print("*",end = " ")
    print()
```

```python
for r in range(1,5):
    for c in range(1,5):
        if r == 1 or r == 4 or c == 1 or c == 4:
            print("*", end = " ")
        else:
            print(" ", end=  " ")
    print()
            
        
```

```python
num = 1
for r in range(1,4):
    for c in range(1,4):
        print(num, end = " ")
        num += 1
    print()
```

```python
num = 10
for r in range(1,4):
    for c in range(1,4):
        print(num, end = " ")
        num -= 1
    print()
```

```python
for r in range(1,4):
    for c in range(1,4):
        if r == 1 or r == 2 or c == 2:
            print("@", end = " ")
        else:
            print("*", end = " ")
    print()
```
