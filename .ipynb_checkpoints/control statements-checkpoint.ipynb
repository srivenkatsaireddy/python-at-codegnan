```python
# print 11,12,13 number is odd or even in nested if statement
n1 = int(input())
n2 = int(input())
n3 = int(input())
if n1 % 2 != 0:
    print(f"{n1} number is odd")
    if n2 % 2 == 0:
        print(f"{n2} number is even")
        if n3 % 2 != 0:
            print(f"{n3} number is odd")
```

```python
# print 11,12,13 number is odd or even in if-else statement
n1 = int(input())
n2 = int(input())
n3 = int(input())
if n1 % 2:
    print(f"{n1} number is odd")
else:
    print(f"{n1} number is even")
if n2 % 2 == 0:
    print(f"{n2} number is even")
else:
    print(f"{n2} number is odd")
if n3 % 2:
    print(f"{n3} number is odd")
else:
    print(f"{n3} number is even")
```

```python
n1, n2, n3, n4= map(int, input().split())
for val in [n1, n2, n3, n4]:
    if val % 2 == 0:
        print(f"{val} is even")
    else:
        print(f"{val} is odd")
else:
    print("program completed")
```

```python
# count number of vowels and consonants
consonant = 0
vowel = 0
for i in ['C','O','D','E','G','N','A','N']:
   if i in 'aeiouAEIOU':
       vowel = vowel + 1
   else:
       consonant = consonant + 1
print(vowel)
print(consonant)
```

```python
# count number of vowels and consonants in combined string{collection of numbers,special characters}
string = input()
c = v = 0
for char in string:
    if char in "aeiou":
       v = v + 1
    elif char.isalpha():
        c = c + 1
print(c, v)
       
```

```python
# count even and odd number in given list of numbers and combined string.
num = input()
e = o = 0
for i in num:
    if i.isdigit():
        if int(i) % 2:
            o = o + 1
        else:
            e = e + 1
print(e,o)
        
```

```python
# seperate the even and odd number in the given list
arr = list(map(int, input().split()))
even = []
odd = []
for num in arr:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print(even)
print(odd)
```

```python
# lst = [1,2,3,4,5,6] o/p: even: [2,4,6] , odd : [1,3,5]
arr = list(map(int, input().split()))
for i in arr:
    if i % 2 == 0:
        print({i})
        print("even number")
    else:
        print({i})
        print("odd number")
```
