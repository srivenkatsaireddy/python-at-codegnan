# Tuple

```python
a = ("sai","pavan",5,6)
print(a)
print(type(a))
```

```python
a = ("welcome","to","codegnan","to")
b = (1,2,3)
print(a.count("to"))
print(a.count("aib"))
print(a+b)
print(a[0])
print(a[::1])
print(a[::-1])
print(a[-6:-1])
print(a.index("codegnan"))
```

```python
t = (10,20,[30])
t[2].append("40")
print(t)
```

```python
a = "hello hyderabad and telangana"
print(a[1])
print(a[:8:1])
print(a[-4:-1])
```

```python
## define empty tuple
t1 = ()
t2 = tuple()
print(type(t1), type(t2))
```

```python
t1 = [3]
t2 = (23)
t3 = (23,)
print(type(t1),type(t2), type(t3))
```
