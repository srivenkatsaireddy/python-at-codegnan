```python
import random
steps=rolls=0
while True:
    input("to roll dice click enter")
    dice_value=random.randint(1,6)
    rolls+=1
    print("current dice value:", dice_value)
    if steps+dice_value<=100:
        steps+=dice_value
    if steps==100:
        print(f"you reached top in {rolls} rolls")
        break
    print(f"For {rolls} rolls you are at {steps} position")    
```
