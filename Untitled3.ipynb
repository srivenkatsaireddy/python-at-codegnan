```python
s = input()

for op in "+-*/":
    if op in s:
        a, b = map(int, s.split(op))
        if op == "+":
            print(a + b)
        elif op == "-":
            print(a - b)
        elif op == "*":
            print(a * b)
        else:
            print(a // b)
        break
```
