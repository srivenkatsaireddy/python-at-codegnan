```python
import pandas as pd
df = pd.read_csv("Housing.csv")
categorical_columns = ['mainroad','guestroom','basement','hotwaterheating', 'airconditioning',
       'parking', 'prefarea', 'furnishingstatus']
def transfer(x):
    if x == 'yes':
        return 1
    else:
        return 0
df['guestroom'] = df['guestroom'].apply(transfer)
df['basement'] = df['basement'].apply(transfer)
df['hotwaterheating'] = df['hotwaterheating'].apply(transfer)
df['airconditioning'] = df['airconditioning'].apply(transfer)
df['parking'] = df['parking'].apply(transfer)
df['prefarea'] = df['prefarea'].apply(transfer)
print(df[["guestroom", "basement", "hotwaterheating", "airconditioning", "prefarea"]])


```

```python
df['furnishingstatus'] = df['furnishingstatus'].apply(transfer)
print(df["furnishingstatus"])
```

```python
def transfer(x):
    if x == 'unfurnished':
        return 0
    elif x== "semi-furnished":
        return 0.5
    else:
        return 1
df["furnishingstatus"]=df["furnishingstatus"].apply(transfer)
df
```
