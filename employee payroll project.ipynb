```python
class Employee:

    def __init__(self, emp_id, name, department, basic_salary):
        self.emp_id = emp_id
        self.name = name
        self.department = department
        self.basic_salary = basic_salary

    def calculate_salary(self):
        hra = self.basic_salary * 0.20
        da = self.basic_salary * 0.10
        tax = self.basic_salary * 0.05

        net_salary = self.basic_salary + hra + da - tax

        return net_salary

    def display(self):
        print("\n-----------------------------")
        print("Employee ID :", self.emp_id)
        print("Name        :", self.name)
        print("Department  :", self.department)
        print("Basic Salary:", self.basic_salary)
        print("Net Salary  :", self.calculate_salary())
        print("-----------------------------")


employees = []


def add_employee():
    emp_id = int(input("Enter Employee ID: "))
    name = input("Enter Name: ")
    department = input("Enter Department: ")
    salary = float(input("Enter Basic Salary: "))

    emp = Employee(emp_id, name, department, salary)

    employees.append(emp)

    print("\nEmployee Added Successfully.\n")


def view_all():
    if len(employees) == 0:
        print("\nNo Employees Found.\n")
    else:
        for emp in employees:
            emp.display()


def search_employee():
    search_id = int(input("Enter Employee ID: "))

    for emp in employees:
        if emp.emp_id == search_id:
            emp.display()
            return

    print("\nEmployee Not Found.\n")


def delete_employee():
    delete_id = int(input("Enter Employee ID to Delete: "))

    for emp in employees:
        if emp.emp_id == delete_id:
            employees.remove(emp)
            print("\nEmployee Deleted Successfully.\n")
            return

    print("\nEmployee Not Found.\n")


while True:

    print("\n========== Employee Payroll System ==========")
    print("1. Add Employee")
    print("2. View All Employees")
    print("3. Search Employee")
    print("4. Delete Employee")
    print("5. Exit")

    choice = input("Enter Your Choice: ")

    if choice == "1":
        add_employee()

    elif choice == "2":
        view_all()

    elif choice == "3":
        search_employee()

    elif choice == "4":
        delete_employee()

    elif choice == "5":
        print("\nThank You...")
        break

    else:
        print("\nInvalid Choice.")
```
