```python
def num(n):
    if n == 0:
        return
    print(n)
    num(n-1)
num(5)
```

```python
def num(n):
    if n == 6:
        return
    print(n)
    num(n+1)
num(1)
```

```python
def fact(n):
    if n == 0 or n == 1:
        return 1
    return n * fact(n-1)
    
print(fact(6))
```

```python
def sum(n):
    if n == 0:
        return 0
    return n + sum(n-1)
print(sum(10))
```

```python
def num(a,b):
    if b == 0:
        return 1
    return a * num(a,b-1)
print(num(3,4))
```

```python
def num(n):
    if n == 0:
        return 0
    print(n)
    num(n-1)
num(10)
```

```python
def even_numbers(n):
    if n == 1:
        return 0
    if n % 2 == 0:
        print(n)
    even_numbers(n-1)
even_numbers(10)
```
